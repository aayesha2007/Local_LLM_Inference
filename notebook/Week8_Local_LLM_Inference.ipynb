```python
import pandas as pd
from pathlib import Path
import ollama
from IPython.display import display

# Project paths
BASE_DIR = Path.cwd()

if BASE_DIR.name == "notebook":
    BASE_DIR = BASE_DIR.parent

DATASET_DIR = BASE_DIR / "dataset"

# Load prompts and generated responses
prompts_df = pd.read_csv(DATASET_DIR / "prompts.csv")
responses_df = pd.read_csv(DATASET_DIR / "llm_responses.csv")
comparison_df = pd.read_csv(DATASET_DIR / "response_comparison.csv")

print("LOCAL LLM INFERENCE PROJECT")
print("Model: Llama 3.2")
print("Total prompts:", len(prompts_df))
print("Successful responses:", (responses_df["status"] == "Success").sum())

display(prompts_df)
display(responses_df[["id", "domain", "prompt_type",
                      "response_length_words", "status"]])
display(comparison_df)
```

```python
# Compare a vague prompt with a clearer prompt

weak_prompt = "Tell me about Python."

improved_prompt = (
    "Explain Python programming to a beginner in five bullet points. "
    "Include one simple example and explain its output."
)

def ask_model(prompt):
    result = ollama.chat(
        model="llama3.2",
        messages=[{"role": "user", "content": prompt}]
    )
    return result["message"]["content"]

weak_response = ask_model(weak_prompt)
improved_response = ask_model(improved_prompt)

print("WEAK PROMPT:")
print(weak_response)

print("\nIMPROVED PROMPT:")
print(improved_response)

print("\nOBSERVATION:")
print(
    "The weak prompt is broad and does not specify the audience or format. "
    "The improved prompt gives clear instructions, a target audience, "
    "a response structure, and an example requirement."
)
```
